```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Problem constants
# -----------------------------
E0_mean = 3000.0        # MeV
alpha = 2.0             # MeV/cm
E_threshold = 800.0     # MeV
sigmas = [100, 200, 300] # MeV
N = 5000                # number of muons to simulate
```

```python
def generate_initial_energies(N, sigma):
    return np.random.normal(E0_mean, sigma, N)


def compute_final_energies(E0_values):
    x = np.random.uniform(0, 1000, len(E0_values))  # depth in cm
    E_exp = E0_values - alpha * x
    return x, E_exp
```

```python
fig, axs = plt.subplots(1, 3, figsize=(15, 4))
for i, sigma in enumerate(sigmas):
    E0 = generate_initial_energies(N, sigma)
    print(E0.shape)
    axs[i].plot(E0, '.', alpha=0.3)
    axs[i].set_title(f"σ_E = {sigma} MeV")
    axs[i].set_xlabel("Muon index")
    axs[i].set_ylabel("Initial energy E0 [MeV]")
    axs[i].set_ylim(2000, 4000)
plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(1, 3, figsize=(15, 4))
for i, sigma in enumerate(sigmas):
    E0 = generate_initial_energies(N, sigma)
    Ef = compute_final_energies(E0)[1]
    print(np.min(Ef))
    print(Ef.shape)
    axs[i].plot(Ef, '.', alpha=0.3,)
    axs[i].set_title(f"σ_E = {sigma} MeV")
    axs[i].set_xlabel("Muon index")
    axs[i].set_ylabel("Initial energy E0 [MeV]")
    axs[i].set_ylim(00, 4000)
plt.tight_layout()
plt.show()
```
